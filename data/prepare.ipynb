# 数据预处理

只读取 data/xb/simscDNAm 下的原始 H5AD；本地原始数据放在 data/raw/。按源数据集 × Subclass 固定种子分层划分 8:2，保留真实零、NaN；训练前追加相同的 1000 个复制区域。所有模型使用这里导出的同一划分。

```python
from pathlib import Path
import sys
PROJECT = next(p for p in (Path.cwd(), *Path.cwd().parents)
    if (p / "model/src/model.py").is_file())
SERVER = sys.platform != "win32"
DATA_ROOT = Path("/home/sccasimp/data/xb/simscDNAm") if SERVER else PROJECT / "data/raw"
OUTPUTS = Path("/home/sccasimp/data/xb/outputs/simscDNAm") if SERVER else PROJECT / "outputs"
DATASET = "liu2023/full"
INPUT_PATHS = [DATA_ROOT / "Liu-2023_nature" / accession / f"{accession}.h5ad"
    for accession in ("GSE213221", "GSE215213", "GSE215342", "GSE230224", "GSE230226")]
OUTPUT_DIR = OUTPUTS / "preprocessing" / DATASET
SEED = 42
TEST_FRACTION = 0.2
```

```python
sys.path.insert(0, str(PROJECT))
from data.preprocessing import prepare_dataset
assert Path(sys.prefix).name.lower() == "simscdnam", sys.executable
for path in INPUT_PATHS:
    path.resolve().relative_to(DATA_ROOT.resolve())
OUTPUT_DIR.resolve().relative_to(OUTPUTS.resolve())
report = prepare_dataset(INPUT_PATHS, OUTPUT_DIR, seed=SEED, test_fraction=TEST_FRACTION)
print(report)
```
